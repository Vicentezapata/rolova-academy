## Ejemplo básico de clasificación de nodos

Para este ejemplo usaremos una capa de tipo **GCN**, propuesta por [Kipf et al. (2017)](https://arxiv.org/abs/1609.02907), que se define como

$$
\mathbf{h}_i^{(t + 1)} = \mathbf{W}^{(t + 1)} \sum_{j \in \mathcal{N}(i) \, \cup \, \{ i \}} \frac{1}{c_{i,j}} \cdot \mathbf{h}_j^{(t)}
$$

donde $\mathbf{W}^{(t + 1)}$ es una matriz entrenable de tamaño `[num_output_features, num_input_features]` y $c_{i,j}$ es un coeficiente fijo de normalización por cada arista.

Pytorch Geometric implementa esta capa mediante la clase [`GCNConv`](https://pytorch-geometric.readthedocs.io/en/latest/modules/nn.html#torch_geometric.nn.conv.GCNConv) y el modo de uso es (casi) idéntico a como usamos las convoluciones de pytorch para imágenes.

````
import torch
from torch.nn import Linear
from torch_geometric.nn import GCNConv

class GCN(torch.nn.Module):
    def __init__(self):
        super(GCN, self).__init__()
        torch.manual_seed(1)
        # creamos nuestra GCN con 3 capas convolucionales
        self.conv1 = GCNConv(dataset.num_features, 4)
        self.conv2 = GCNConv(4, 4)
        self.conv3 = GCNConv(4, 2)
        # finalmente nuestra capa de clasificación a num_clases
        self.classifier = Linear(2, dataset.num_classes)

    def forward(self, x, edge_index):
        h = self.conv1(x, edge_index)
        h = h.tanh()
        h = self.conv2(h, edge_index)
        h = h.tanh()
        h = self.conv3(h, edge_index)
        h = h.tanh()
        out = self.classifier(h)
        return out, h
````

Ahora podemos visualizar nuestra GCN antes de entrenar.
Es interesante ver que a pesar de tener pesos **completamente aleatorios** sin ningún tipo de entrenamiento, la forma en que se definen las convoluciones de grafos introduce un bias inductivo poderoso, generando embeddings que se asemejan a la estructura de comunidad del grafo al asignar embeddings similares a nodos cercanos en el grafo.

````
import numpy as np
import matplotlib.pyplot as plt
from sklearn.manifold import TSNE
import torch

WIDTH_PX, HEIGHT_PX, DPI = 450, 450, 100
_FIGSIZE = (WIDTH_PX / DPI, HEIGHT_PX / DPI)

def _fig_to_rgb_array(fig):
    # para dibujar a imagen RGB: ndarray (H, W, 3)
    fig.canvas.draw()
    w, h = fig.canvas.get_width_height()
    buf = np.frombuffer(fig.canvas.buffer_rgba(), dtype=np.uint8).reshape(h, w, 4)
    rgb = buf[..., :3].copy()
    plt.close(fig)
    return rgb

def visualize_embedding(
    h,
    color,
    epoch=None,
    loss=None,
    tsne: bool = False,
    return_frame: bool = False,
    point_size: int = 140,
    cmap: str = "Set2",
    dpi: int = 120,   # kept for backward compat; ignored because size is hard-coded
):
    # normalizar inputs
    h_np = h.detach().cpu().numpy() if torch.is_tensor(h) else np.asarray(h)
    color_np = color.detach().cpu().numpy() if torch.is_tensor(color) else np.asarray(color)

    # elegir representacion 2D segun corresponda (normal o TSNE)
    use_tsne = tsne or (h_np.ndim != 2) or (h_np.shape[1] != 2)
    xy = TSNE(n_components=2, init="pca").fit_transform(h_np) if use_tsne else h_np[:, :2]

    # Dibujar
    fig = plt.figure(figsize=_FIGSIZE, dpi=DPI)
    ax = fig.add_subplot(111)
    ax.scatter(
        xy[:, 0], xy[:, 1],
        s=point_size if not use_tsne else max(70, point_size // 2),
        c=color_np, cmap=cmap
    )
    ax.set_xticks([]); ax.set_yticks([])
    if epoch is not None and loss is not None:
        loss_val = float(loss.item() if hasattr(loss, "item") else loss)
        ax.set_xlabel(f"Epoch: {epoch}  |  Loss: {loss_val:.4f}", fontsize=14)
    fig.tight_layout(pad=0)

    if return_frame:
        return _fig_to_rgb_array(fig)
    else:
        plt.show()
        return None
````

````
model = GCN()
model.eval()

with torch.no_grad():
    _, h = model(data.x, data.edge_index)
print(f'Embedding shape: {list(h.shape)}')

visualize_embedding(h, color=data.y)
````

Ahora entrenemos nuestra **GCN** para ver si es capaz de generar una mejor separación de las clases

````
from IPython.display import Javascript, Image, display
import imageio.v2 as imageio
import torch
import pathlib

# Aquí definimos nuestro modelo, función de pérdida y optimizador
model = GCN()
criterion = torch.nn.CrossEntropyLoss()
optimizer = torch.optim.Adam(model.parameters(), lr=0.01)

def train_step(data):
    optimizer.zero_grad()  # resetear gradientes
    out, h = model(data.x, data.edge_index)  # forward pass.
    loss = criterion(out[data.train_mask], data.y[data.train_mask])  # calcular pérdida basados en los nodos de training
    loss.backward()  # backward pass.
    optimizer.step()  # actualización de pesos (guiados por el gradiente).
    return loss, h

# Parametros para capturar a GIF
every_n = 10          # capturar un frame cada N epocas
frame_duration = 0.30 # duracion en segundos de cada frame en el GIF
frames = []

# Frame inicial (epoca 0, antes de entrenar)
with torch.no_grad():
    _, h0 = model(data.x, data.edge_index)
frame0 = visualize_embedding(h0, color=data.y, epoch=0, loss=0.0, return_frame=True)
frames.append(frame0)

# Loop de entrenamiento
for epoch in range(1, 801):
    loss, h = train_step(data)
    if epoch % every_n == 0:
        frame = visualize_embedding(h, color=data.y, epoch=epoch, loss=loss, return_frame=True)
        frames.append(frame)

# Guardamos el GIF en el directorio actual (compatible con Windows y Colab)
gif_path = pathlib.Path("gcn_training.gif")
imageio.mimsave(str(gif_path), frames, duration=frame_duration, loop=0)  # 0 = loop infinito
display(Image(str(gif_path)))
````

Luego de entrenar vemos que logra separar mejor las clases (colores) aunque no es perfecto, lo cual es esperable dado el grafo de entrada, que no tiene informacion muy discriminativa para nodos intermedios (que por conectividad podian asignarse a cualquiera de las clases indistintamente, en cierta forma)

## Ejemplo más avanzado de clasificación de nodos

En este ejemplo usaremos un grafo más complejo, llamado "Cora" que contiene información de publicaciones (papers), cada uno con features indicando la presencia o ausencia de ciertas palabras clave, y una clase (categoría del paper). El grafo se construye mediante las citas de un paper referenciando a otro.

````
from torch_geometric.datasets import Planetoid
from torch_geometric.transforms import NormalizeFeatures

dataset = Planetoid(root='data/Planetoid', name='Cora', transform=NormalizeFeatures())

print(f'Number of graphs: {len(dataset)}')
print(f'Number of features: {dataset.num_features}')
print(f'Number of classes: {dataset.num_classes}')

data = dataset[0] # Obtener el primer (y unico) grafo, como objeto de clase Data
print(f'Number of nodes: {data.num_nodes}')
print(f'Number of edges: {data.num_edges}')
print(f'Average node degree: {data.num_edges / data.num_nodes:.2f}')
print(f'Contains isolated nodes: {data.has_isolated_nodes()}')
print(f'Contains self-loops: {data.has_self_loops()}')
print(f'Is undirected: {data.is_undirected()}')
````

````
def train(some_model, some_data):
      some_model.train()
      optimizer.zero_grad()  # resetear gradientes
      out = some_model(some_data.x)  # forward pass.
      loss = criterion(out[some_data.train_mask], some_data.y[some_data.train_mask])  # calcular pérdida basados en los nodos de training
      loss.backward()  # backward pass.
      optimizer.step()  # actualización de pesos (guiados por el gradiente).
      return loss

def test(some_model, some_data):
      some_model.eval()
      out = some_model(some_data.x)
      pred = out.argmax(dim=1)  # Usar la clase con probabilidad mas alta
      test_correct = pred[some_data.test_mask] == some_data.y[some_data.test_mask]  # Comparar con Ground Truth
      test_acc = int(test_correct.sum()) / int(some_data.test_mask.sum())  # correctos dividido totales
      return test_acc
````

Para tener un punto de comparacion, a continuación crearemos un clasificador tipo MLP que sea capaz de distinguir entre las distintas categorías de papers a partir de los features (ignorando el gráfico)

````
import torch
from torch.nn import Linear
import torch.nn.functional as F

class MLP(torch.nn.Module):
    def __init__(self, hidden_channels):
        super(MLP, self).__init__()
        torch.manual_seed(12345)
        self.lin1 = Linear(dataset.num_features, hidden_channels)
        self.lin2 = Linear(hidden_channels, dataset.num_classes)

    def forward(self, x):
        x = self.lin1(x)
        x = x.relu()
        x = F.dropout(x, p=0.5, training=self.training)
        x = self.lin2(x)
        return x

# Aquí creamos nuestro modelo, función de pérdida y optimizador
model = MLP(hidden_channels=16)
criterion = torch.nn.CrossEntropyLoss()  # Define loss criterion.
optimizer = torch.optim.Adam(model.parameters(), lr=0.01, weight_decay=5e-4)  # Define optimizer.

for epoch in range(1, 201):
    loss = train(model, data)
    print(f'Epoch: {epoch:03d}, Loss: {loss:.4f}')
````

````
test_acc = test(model, data)
print(f'Test Accuracy: {test_acc:.4f}')
````

## Ejemplo de clasificación de Grafos

En algunos casos, en lugar de querer clasificar nodos, nos interesa clasificar grafos completos. A continuación haremos clasificación de moléculas (a partir del grafo que representa los enlaces entre los átomos que la componen).

Existen muchos datasets (algunos enormes) con este tipo de información. A continuación usaremos el **MUTAG dataset**, que es más pequeño. Este dataset clasifica moléculas según si inhiben o no la replicación del virus VIH.

Si inspeccionamos el dataset, veremos que tiene 188 grafos. Podemos inspeccionar cada uno por separado (en el ejemplo, revisaremos solo uno, lo cual se refleja en la linea data = dataset[15], es decir, tomamos el grafo de indice 15)

````
import torch
from torch_geometric.datasets import TUDataset

dataset = TUDataset(root='data/TUDataset', name='MUTAG')

print(f'Number of graphs: {len(dataset)}')
print(f'Number of features: {dataset.num_features}')
print(f'Number of classes: {dataset.num_classes}')


print('-'*50+"\nInformation of the Graph #15")
data = dataset[15] # obtengamos uno de los 188 grafos
print(f'Number of nodes: {data.num_nodes}')
print(f'Number of edges: {data.num_edges}')
print(f'Average node degree: {data.num_edges / data.num_nodes:.2f}')
print(f'Contains isolated nodes: {data.has_isolated_nodes()}')
print(f'Contains self-loops: {data.has_self_loops()}')
print(f'Is undirected: {data.is_undirected()}')
````

PyTorch Geometric entrega funcionalidades para desordenar de manera aleatoria el orden de los grafos. Usaremos esa herramienta para dividir nuestro dataset en 150 para training y los 38 restantes para test:

````
from torch_geometric.loader import DataLoader

torch.manual_seed(12345)
dataset = dataset.shuffle()

train_dataset = dataset[:150]
test_dataset = dataset[150:]

print(f'Number of training graphs: {len(train_dataset)}')
print(f'Number of test graphs: {len(test_dataset)}')

train_loader = DataLoader(train_dataset, batch_size=64, shuffle=True)
test_loader = DataLoader(test_dataset, batch_size=64, shuffle=False)
````

Ahora crearemos nuestra GNN compuesto de tres convoluciones de grafos y un clasificador lineal final.

````
from torch.nn import Linear
import torch.nn.functional as F
from torch_geometric.nn import GCNConv
from torch_geometric.nn import global_mean_pool


class GCN(torch.nn.Module):
    def __init__(self, hidden_channels):
        super(GCN, self).__init__()
        torch.manual_seed(12345)
        self.conv1 = GCNConv(dataset.num_node_features, hidden_channels)
        self.conv2 = GCNConv(hidden_channels, hidden_channels)
        self.conv3 = GCNConv(hidden_channels, hidden_channels)
        self.lin = Linear(hidden_channels, dataset.num_classes)

    def forward(self, x, edge_index, batch):
        # 1. Calcular los embeddings de los nodos
        x = self.conv1(x, edge_index)
        x = x.relu()
        x = self.conv2(x, edge_index)
        x = x.relu()
        x = self.conv3(x, edge_index)

        # 2. Leer los embeddings de todos los nodos (Readout)
        x = global_mean_pool(x, batch)  # [batch_size, hidden_channels]

        # 3. Aplicar dropout y un clasificador lineal
        x = F.dropout(x, p=0.5, training=self.training)
        x = self.lin(x)

        return x
````

Definimos los metodos de train y test, identicos a los que usariamos en una red convolucional como las que hemos visto en clases anteriores.

````
def train(loader, model):
    model.train()
    for data in loader:
         out = model(data.x, data.edge_index, data.batch)
         loss = criterion(out, data.y)
         loss.backward()
         optimizer.step()
         optimizer.zero_grad()

def test(loader, model):
     model.eval()
     correct = 0
     for data in loader:
         out = model(data.x, data.edge_index, data.batch)
         pred = out.argmax(dim=1)  # Usar la clase con mayor probabilidad
         correct += int((pred == data.y).sum())  # Comparar con las etiquetas reales
     return correct / len(loader.dataset)  # Retornar el porcentaje de clasificaciones correctas

model = GCN(hidden_channels=64)
optimizer = torch.optim.Adam(model.parameters(), lr=0.01)
criterion = torch.nn.CrossEntropyLoss()

for epoch in range(1, 201):
    train(train_loader, model)
    train_acc = test(train_loader, model)
    print(f'Epoch: {epoch:03d}, Train Acc: {train_acc:.4f}')
````

````
test_acc = test(test_loader, model)
print(f'Test Acc: {test_acc:.4f}')
````

Vemos que el modelo logra un buen rendimiento a pesar de usar una red muy sencilla y muy pocos datos

## (Material Opcional) Probando la técnica de Morris et al.

Algunos papers ([Xu et al. (2018)](https://arxiv.org/abs/1810.00826), [Morris et al. (2018)](https://arxiv.org/abs/1810.02244)), muestran que la normalización de vecinos perjudican la expresividad de las GNN para distinguir ciertos tipos de estructuras de grafos.


[Morris et al. (2018)](https://arxiv.org/abs/1810.02244) quita la normalización en la etapa de agregación de mensajes y agrega una skip-connection para permitir que la red preserve la información original de cada nodo:

$$
\mathbf{h}_i^{(t+1)} = \mathbf{W}^{(t + 1)}_A \mathbf{h}_i^{(t)} + \mathbf{W}^{(t + 1)}_B \sum_{j \in \mathcal{N}(i)} \mathbf{h}_j^{(t)}
$$

Para usar esta capa basta con usar su implementación en la clase [`GraphConv`](https://pytorch-geometric.readthedocs.io/en/latest/modules/nn.html#torch_geometric.nn.conv.GraphConv) de PyTorch Geometric en lugar de la GCNConv que usamos *anteriormente*.

````
from torch_geometric.nn import GraphConv


class GNN(torch.nn.Module):
    def __init__(self, hidden_channels):
        super(GNN, self).__init__()
        torch.manual_seed(12345)
        self.conv1 = GraphConv(dataset.num_node_features, hidden_channels)
        self.conv2 = GraphConv(hidden_channels, hidden_channels)
        self.conv3 = GraphConv(hidden_channels, hidden_channels)
        self.lin = Linear(hidden_channels, dataset.num_classes)

    def forward(self, x, edge_index, batch):
        x = self.conv1(x, edge_index)
        x = x.relu()
        x = self.conv2(x, edge_index)
        x = x.relu()
        x = self.conv3(x, edge_index)

        x = global_mean_pool(x, batch)

        x = F.dropout(x, p=0.5, training=self.training)
        x = self.lin(x)

        return x
````

````
model = GNN(hidden_channels=64)
print(model)
optimizer = torch.optim.Adam(model.parameters(), lr=0.01)

for epoch in range(1, 201):
    train(train_loader, model)
    train_acc = test(train_loader, model)
    print(f'Epoch: {epoch:03d}, Train Acc: {train_acc:.4f}')
````

````
test_acc = test(test_loader, model)
print(f'Test Acc: {test_acc:.4f}')
````

Con esto vemos que el pequenio cambio que introduce Morris et al. produce una mejora importante en el resultado, lo cual se explica por el hecho de que en estos casos permite a la red diferenciar la cantidad de aristas que llegan a cada nodo (una GCN comun pierde esa informacion al preservar solo el valor promedio)

**¿Por qué?** La operación matemática de una capa GCN es:

$$\mathbf{h}_i^{(t+1)} = \mathbf{W}^{(t+1)} \sum_{j \in \mathcal{N}(i) \cup \{i\}} \frac{1}{c_{ij}} \cdot \mathbf{h}_j^{(t)}$$

Aunque la matriz $\mathbf{W}$ tenga valores **aleatorios**, el proceso de **agregar/promediar los features de los nodos vecinos** sigue estrictamente la estructura de las aristas del grafo.

**Analogía sencilla:**
> Imagina que cada nodo "escucha" a sus vecinos y promedia lo que escucha. Si el nodo A tiene 5 vecinos que pertenecen al grupo rojo, su embedding quedará muy influenciado por el grupo rojo — incluso antes de entrenar nada.

**Concepto clave — Inductive Bias (sesgo inductivo):**
- La GCN tiene un sesgo *natural* hacia la topología del grafo.
- Nodos **cercanos** en el grafo → reciben mensajes similares → producen embeddings similares.
- Nodos **lejanos** en el grafo → reciben mensajes distintos → producen embeddings distintos.

Por eso, aunque los pesos sean aleatorios, los nodos del mismo "grupo" (comunidad) ya aparecen agrupados en el espacio de embeddings.

### 💡 Explicación — Actividad 2: Problema con nodos sin aristas entrantes

#### Problema identificado
En la fórmula de la GCN, cada nodo **agrega los features de sus vecinos entrantes**:

$$\mathbf{h}_i^{(t+1)} = \mathbf{W} \sum_{j \in \mathcal{N}(i)} \frac{1}{c_{ij}} \cdot \mathbf{h}_j^{(t)}$$

Si un nodo **no tiene ninguna arista entrante** ($\mathcal{N}(i) = \emptyset$), la suma es **cero**, y el resultado es:
$$\mathbf{h}_i^{(t+1)} = \mathbf{W} \cdot \mathbf{0} = \mathbf{0}$$

El nodo pierde toda su información original de features → su representación queda sin sentido.

#### Solución: Self-Loops (AutoConexiones)
Se añade una **arista artificial de cada nodo hacia sí mismo**. Esto garantiza que:
- $\mathcal{N}(i)$ siempre incluye al menos el nodo $i$
- La suma nunca es vacía → el nodo siempre "hereda" al menos sus propios features

```
Antes:   A → B → C        (B solo recibe desde A, C se queda sin info entrante)
Después: A → B → C        (igual que antes)
         ↙  ↙  ↙          (más self-loops: A→A, B→B, C→C)
```

> **En PyG:** `GCNConv` aplica `add_self_loops=True` **automáticamente** por defecto.

El **grado** de un nodo es simplemente **cuántos vecinos tiene** (cuántas aristas están conectadas a él).

#### ¿Por qué `num_edges / num_nodes` da el grado promedio?

En PyTorch Geometric, los grafos **no dirigidos** almacenan cada conexión **dos veces**:
- La arista A-B se guarda como `A → B` **y** como `B → A`

Esto significa que `num_edges = 2 × aristas_reales`.

Si sumamos los grados de todos los nodos, obtenemos `2 × aristas_reales` (ya que cada arista "toca" exactamente 2 nodos). Por lo tanto:

$$\text{grado\_promedio} = \frac{\sum \text{grados}}{N} = \frac{2 \times \text{aristas}}{N} = \frac{\texttt{num\_edges}}{\texttt{num\_nodes}}$$

### 💡 Explicación — Actividad 6: Campo receptivo de una GCN

#### Concepto clave: "Campo Receptivo" (Receptive Field)

En redes convolucionales clásicas (para imágenes), una CNN con $k$ capas puede "ver" una región de $(2k+1) \times (2k+1)$ píxeles de la imagen original. En GNNs ocurre algo análogo:

> **Con $k$ capas GCN, un nodo puede agregar información de nodos que están a lo sumo $k$ saltos de distancia.**

#### Visualización del flujo de información (3 capas):

```
Capa 1: A agrega sus vecinos directos     (nodos a 1 salto)
Capa 2: A agrega vecinos de sus vecinos   (nodos a 2 saltos)
Capa 3: A agrega vecinos de los anteriores (nodos a 3 saltos)
                                           ↑
                                     LÍMITE MÁXIMO
```

#### Aplicado al problema:

- **A y B están separados por 4 saltos**
- **El modelo tiene 3 capas → alcance máximo = 3 saltos**
- **3 < 4 → La información de B NO llega a A**

```
B ──── X ──── Y ──── Z ──── A
  hop4    hop3    hop2    hop1
                          ^
                     Aquí llega
                    la capa 3 de A
```

#### Respuesta: **No**, la clasificación de A no considera información de B.

> **Reflexión importante:** Añadir más capas no siempre es bueno. Demasiadas capas pueden causar el problema de **"over-smoothing"**: todos los nodos terminan con embeddings muy similares (indistinguibles). Por eso, en GNNs se suele usar pocas capas (2-4).
