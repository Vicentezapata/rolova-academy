# Descenso del Gradiente en un Modelo Lineal

## Modelo

Sea un modelo lineal con una sola entrada y una salida:

$$
\hat{y} = wx + b
$$

donde:
- \( x \) es la entrada,
- \( w \) es el peso,
- \( b \) es el sesgo (bias),
- \( \hat{y} \) es la predicción del modelo.

## Pérdida

Usamos el error cuadrático medio como función de pérdida para un solo ejemplo:

$$
\mathcal{L} = \frac{1}{2}(\hat{y} - y)^2 = \frac{1}{2}(wx + b - y)^2
$$

donde \( y \) es la etiqueta verdadera.

## Derivadas (Gradientes)

Queremos minimizar la pérdida derivando respecto a los parámetros:

### Gradiente con respecto a \( w \):

$$
\frac{\partial \mathcal{L}}{\partial w} = (wx + b - y) \cdot x
$$

### Gradiente con respecto a \( b \):

$$
\frac{\partial \mathcal{L}}{\partial b} = (wx + b - y)
$$

## Regla de Actualización (Descenso del Gradiente)

Dado un learning rate \( $\eta$ \), actualizamos así:

$$
w \leftarrow w - \eta \cdot \frac{\partial \mathcal{L}}{\partial w}
$$

$$
b \leftarrow b - \eta \cdot \frac{\partial \mathcal{L}}{\partial b}
$$

## Repetimos esto para múltiples ejemplos y múltiples épocas.


```python
import numpy as np
import matplotlib.pyplot as plt
from matplotlib.animation import FuncAnimation
from IPython.display import HTML

# === Parámetros globales ===
n_points = 30           # <<-- Cambia este valor para usar más o menos datos
lr = 0.01
n_epochs = 50
noise_std = 0.5         # Desviación estándar del ruido
true_w, true_b = 2.0, 0.0  # Parámetros reales

# === Generar datos sintéticos ===
np.random.seed(0)
x_train = np.linspace(1, 5, n_points)
y_train = true_w * x_train + true_b + np.random.normal(0, noise_std, size=x_train.shape)

# === Inicializar pesos ===
w = 0.0
b = 0.0

# === Guardar historia ===
history = {"w": [], "b": [], "loss": []}

# === Simular entrenamiento via Descenso del Gradiente===
for _ in range(n_epochs):
    y_pred = w * x_train + b
    error = y_pred - y_train
    dw = np.mean(error * x_train) # calculo gradiente
    db = np.mean(error)           # calculo gradiente
    w -= lr * dw
    b -= lr * db
    loss = np.mean((w * x_train + b - y_train) ** 2)
    history["w"].append(w)
    history["b"].append(b)
    history["loss"].append(loss)


# ----------- De aquí en adelante es sólo para visualización ---------------
# === Crear figura ===
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(11, 4))
title = fig.suptitle("")


# Panel 1: Ajuste
ax1.set_xlim(x_train.min() - 1, x_train.max() + 1)
ax1.set_ylim(y_train.min() - 2, y_train.max() + 2)
sc = ax1.scatter(x_train, y_train, label="Datos con ruido")
line, = ax1.plot([], [], 'r-', lw=2, label="Regresión")
ax1.plot(x_train, true_w * x_train + true_b, 'g--', label="Línea real (sin ruido)")
ax1.set_title("Ajuste de la recta")
ax1.set_xlabel("x")
ax1.set_ylabel("y")
ax1.legend()

# Panel 2: Pérdida
ax2.set_xlim(0, n_epochs)
ax2.set_ylim(0, max(history["loss"]) + 1)
loss_line, = ax2.plot([], [], 'b-')
loss_points, = ax2.plot([], [], 'ro', markersize=3)
ax2.set_title("Error cuadrático medio")
ax2.set_xlabel("Época")
ax2.set_ylabel("Loss")

# === Función de animación ===
def animate(i):
    w = history["w"][i]
    b = history["b"][i]
    y_pred = w * x_train + b
    error = y_pred - y_train
    dw = np.mean(error * x_train)
    db = np.mean(error)

    line.set_data(x_train, y_pred)
    loss_line.set_data(range(i+1), history["loss"][:i+1])
    loss_points.set_data(range(i+1), history["loss"][:i+1])

    title.set_text(f"Epoch {i+1} | w = {w:.2f}, b = {b:.2f} | dL/dw = {dw:.2f}, dL/db = {db:.2f}")
    return line, loss_line, loss_points, title

ani = FuncAnimation(fig, animate, frames=n_epochs, interval=200)
HTML(ani.to_jshtml())

```
