## 📚 Contexto: Pipeline completo de Deep Fake

Este notebook es la **etapa final** del pipeline. Toma los archivos generados en los notebooks
anteriores y los combina para producir el DeepFake final.

### Pipeline completo de 3 etapas

```
┌─────────────────────────────────────────────────────────────────┐
│  ETAPA 1: Clonación de voz (notebook Audio)                    │
│  Tortoise TTS → generated.wav                                  │
└──────────────────────────┬──────────────────────────────────────┘
                           │
┌──────────────────────────▼──────────────────────────────────────┐
│  ETAPA 2: Swap de cara (notebook Video)                        │
│  ROOP-FLOYD → video.mp4                                        │
└──────────────────────────┬──────────────────────────────────────┘
                           │
┌──────────────────────────▼──────────────────────────────────────┐
│  ETAPA 3: Fusión (este notebook)                               │
│  ffmpeg merge → merged.mp4  (Deep Fake final)                  │
└─────────────────────────────────────────────────────────────────┘
```

### ¿Por qué es necesario sincronizar audio y video?

El modelo de voz (Tortoise) genera audio de longitud variable. El modelo de video (ROOP-FLOYD)
produce un video de longitud fija (la del original). Para que el DeepFake sea coherente:

1. Se ajusta el **frame rate** del video si es necesario (celdas ffmpeg de frame rate)
2. Se añade **silencio al inicio** del audio si hay desfase temporal
3. Se rellena con **silencio al final** para que audio ≥ duración del video
4. Finalmente `ffmpeg -shortest` corta ambos al mínimo común
