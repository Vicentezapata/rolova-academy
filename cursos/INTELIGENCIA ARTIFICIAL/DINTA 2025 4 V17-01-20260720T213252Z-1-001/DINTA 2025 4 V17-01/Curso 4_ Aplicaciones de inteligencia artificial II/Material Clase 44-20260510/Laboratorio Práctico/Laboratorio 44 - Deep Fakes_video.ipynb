### ¿Cómo funciona ROOP-FLOYD?

ROOP-FLOYD implementa un pipeline de **face-swap** en dos etapas:

```
Video original
    ↓
  [1] Detección de cara (RetinaFace)
    ↓
  [2] Extracción de embedding de cara fuente (ArcFace)
    ↓
  [3] Swap de identidad (InsightFace inswapper)
    ↓
  [4] Mejora de calidad (GFPGAN face restoration)
    ↓
  Video con cara intercambiada
```

### Instrucciones para la imagen de referencia

Al ejecutar `run.py`, ROOP-FLOYD levanta una interfaz Gradio. En ella:
1. **Source image**: subir la fotografía de la cara a insertar (ej. Obama en otra pose/foto)
2. **Target video**: subir el video que será procesado
3. Hacer click en **"Start"** y esperar el procesamiento
4. Descargar el video generado y guardarlo localmente como `video.mp4`
